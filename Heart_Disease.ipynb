Uploading data set and chrcking for missing values

```python
import pandas as pd

# Load the dataset
file_path = '/content/heart.csv'
heart_data = pd.read_csv(file_path)

# Display the first few rows of the dataset
heart_data.head()
```

```python
# Check for missing values in the dataset
missing_values = heart_data.isnull().sum()

missing_values
```

Data Analysis

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Summary statistics for numerical features
summary_statistics = heart_data.describe()

# Plotting distributions of numerical features
plt.figure(figsize=(16, 12))

for i, col in enumerate(['age', 'trestbps', 'chol', 'thalach', 'oldpeak'], start=1):
    plt.subplot(2, 3, i)
    sns.histplot(heart_data[col], kde=True, bins=30)
    plt.title(f'Distribution of {col}')

plt.tight_layout()
plt.show()

summary_statistics
```

Categorical features vs Target Insights

```python
plt.figure(figsize=(20, 12))

# Relationship between categorical features and target
categorical_features = ['sex', 'cp', 'fbs', 'restecg', 'exang', 'slope', 'ca', 'thal']
for i, col in enumerate(categorical_features, start=1):
    plt.subplot(2, 4, i)
    sns.countplot(x=col, hue='target', data=heart_data)
    plt.title(f'{col} vs Target')

plt.tight_layout()
plt.show()

```

Continuous features vs Target Variable

```python
# Relationship between continuous features and target using boxplots
plt.figure(figsize=(20, 12))
continuous_features = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak']
for i, col in enumerate(continuous_features, start=1):
    plt.subplot(2, 3, i)
    sns.boxplot(x='target', y=col, data=heart_data)
    plt.title(f'Target vs {col}')

plt.tight_layout()
plt.show()
```

Data preprocessing

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Separating features and target variable
X = heart_data.drop('target', axis=1)
y = heart_data['target']

# Scaling numerical features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Splitting the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Display the shape of the training and testing sets to confirm
(X_train.shape, X_test.shape, y_train.shape, y_test.shape)
```

Model Selection and training

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score

# Initialize the models
models = {
    "Logistic Regression": LogisticRegression(),
    "K-Nearest Neighbors": KNeighborsClassifier(),
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest": RandomForestClassifier(),
    "Gradient Boosting": GradientBoostingClassifier()
}

# Dictionary to hold accuracy scores
accuracy_scores = {}

# Train and evaluate each model
for name, model in models.items():
    model.fit(X_train, y_train)  # Train the model
    y_pred = model.predict(X_test)  # Make predictions on the test set
    accuracy_scores[name] = accuracy_score(y_test, y_pred)  # Calculate accuracy

accuracy_scores
```

Model Evaluation

Cross Validation

```python
from sklearn.model_selection import cross_val_score


# Define your models in a dictionary
models = {
    "Logistic Regression": LogisticRegression(),
    "K-Nearest Neighbors": KNeighborsClassifier(),
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest": RandomForestClassifier(),
    "Gradient Boosting": GradientBoostingClassifier()
}


# Loop through the models and perform cross-validation
for name, model in models.items():
    scores = cross_val_score(model, X_scaled, y, cv=5, scoring='accuracy')
    print(f"{name} Accuracy: {scores.mean() * 100:.2f}% (+/- {scores.std() * 100:.2f}%)")


```

Hyperparameter Tuning of the models

```python
from sklearn.model_selection import GridSearchCV

# Define hyperparameter grids for each model
param_grids = {
    "Logistic Regression": {'C': [0.01, 0.1, 1, 10, 100]},
    "K-Nearest Neighbors": {'n_neighbors': [3, 5, 7, 10], 'metric': ['euclidean', 'manhattan']},
    "Decision Tree": {'max_depth': [None, 10, 20, 30], 'min_samples_leaf': [1, 2, 4]},
    "Random Forest": {'n_estimators': [100, 200], 'max_depth': [None, 10, 20]},
    "Gradient Boosting": {'n_estimators': [100, 200], 'learning_rate': [0.01, 0.1, 0.2]}
}

# Perform hyperparameter tuning
for name, grid in param_grids.items():
    grid_search = GridSearchCV(models[name], grid, cv=5, scoring='accuracy', n_jobs=-1)
    grid_search.fit(X_scaled, y)
    print(f"{name}: Best Parameters = {grid_search.best_params_}, Accuracy = {grid_search.best_score_:.3f}")

```

Evaluating model performance with different metrics

```python
from sklearn.metrics import precision_score, recall_score, f1_score, roc_auc_score

def evaluate_model(model, X_test, y_test):
    y_pred = model.predict(X_test)
    # Adjust for classifiers that do not support predict_proba
    try:
        y_prob = model.predict_proba(X_test)[:, 1]  # Probabilities for the positive class
        roc_auc = roc_auc_score(y_test, y_prob)
    except AttributeError:
        roc_auc = "N/A"

    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)

    return precision, recall, f1, roc_auc

# Assuming 'models' is your dictionary of trained models
model_metrics = {}

for name, model in models.items():
    metrics = evaluate_model(model, X_test, y_test)  # Use the original X_test here
    model_metrics[name] = metrics

for name, metrics in model_metrics.items():
    print(f"{name}: Precision={metrics[0]:.2f}, Recall={metrics[1]:.2f}, F1={metrics[2]:.2f}, ROC-AUC={metrics[3]}")

```

Confusion Matrix

```python
from sklearn.metrics import confusion_matrix, accuracy_score
import seaborn as sns
import matplotlib.pyplot as plt

# Assuming 'models' is your dictionary of trained models
random_forest_model = models["Random Forest"]
gradient_boosting_model = models["Gradient Boosting"]

# Generate predictions
y_pred_rf = random_forest_model.predict(X_test)
y_pred_gb = gradient_boosting_model.predict(X_test)

# Compute confusion matrices
cm_rf = confusion_matrix(y_test, y_pred_rf)
cm_gb = confusion_matrix(y_test, y_pred_gb)

# Function to plot confusion matrix
def plot_confusion_matrix(cm, model_name):
    plt.figure(figsize=(5,5))
    sns.heatmap(cm, annot=True, fmt="d", cmap='Blues', square=True, cbar=False)
    plt.xlabel('Predicted Label')
    plt.ylabel('True Label')
    plt.title(f'Confusion Matrix for {model_name}')
    plt.show()

# Plot confusion matrices for Random Forest and Gradient Boosting
plot_confusion_matrix(cm_rf, "Random Forest")
plot_confusion_matrix(cm_gb, "Gradient Boosting")


```
